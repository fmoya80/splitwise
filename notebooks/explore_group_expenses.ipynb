# Exploración de gastos de un grupo\n
\n
Antes de ejecutar este notebook, extrae el grupo deseado desde la raíz del proyecto: `python main.py --select-group --limit 50`. El archivo local `data/raw/sample_expenses.json` no se sube a Git.

```python
import json
from pathlib import Path

import pandas as pd

raw_path = Path.cwd().parent / 'data/raw/sample_expenses.json'
if not raw_path.exists():
    raw_path = Path.cwd() / 'data/raw/sample_expenses.json'

with raw_path.open(encoding='utf-8') as file:
    expenses_response = json.load(file)

expenses_df = pd.json_normalize(expenses_response.get('expenses', []))
expenses_df.head()
```
